```r
library(rbounds)
library(sf)
library(tidyr)
library(dplyr)
```

```r
df_matched_fixed <- st_drop_geometry(st_read("./df_matched_fixed.gpkg"))
df_matched_fixed <- df_matched_fixed[order(df_matched_fixed$pair_id), ]
```

```r
treated_temps <- df_matched_fixed$temp[df_matched_fixed$type == 1]
control_temps <- df_matched_fixed$temp[df_matched_fixed$type == 0]
```

```r
results <- psens(x = control_temps, y = treated_temps, Gamma = 3, GammaInc = 0.1)

print(results)
```
