```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
base = (-1.0116, -1.438, -0.585)
step_1 = (-0.8332, -1.176, -0.490)
step_2 = (-0.6352, -1.1371, -0.1334)

df = pd.DataFrame((base, step_1, step_2), columns=["coefficient", "err_min", "err_max"], index=["Baseline", "Covariates", "Spat ial correlation"],)
```

```python
fig, ax = plt.subplots(figsize=(5, 6))
sns.pointplot(data=df.transpose(), orient="h", ax=ax, linestyle="none")
_ = ax.set_xlabel("Coefficient value")
```
